# Conditional Statements
- these are the codes which execute based on some condition

```python
if
elif
else
```

```python
# if
```

```python
syntax
if condition:
    code1
    .
    .
    codeN
```

```python
# Condition = True
a = 10
b = 20
if a==10:# True
    print("value of a is 10")
print("Out of the condtional block")
```

```python
# Condition = False
a = 10
b = 20
if a==99: # false
    print("value of a is 10")
print("Out of the condtional block")
```

```python
# else - when we 2 conditons
```

```python
syntax
if condition:
    code1
    code2
else:
    code3
    code4
```

```python
# Condition = True
a = 10
b = 20
if a==10:# True
    print("value of a is 10")
else:
    print("Value of a is 9999")
print("Out of the condtional block")
```

```python
# Condition = False
a = 9999
b = 20
if a==10:# False
    print("value of a is 10")
else:
    print("Value of a is 9999")
print("Out of the condtional block")
```

```python
# elif - when we have more than 2 conditions 
```

```python
syntax
if condition 1:
    code1
    code2
elif Condition 2:
    code3
    code4
else:
    code5
    code6
```

```python
# if Condition = True
a = 10
b = 20
if a==10:# True
    print("value of a is 10")
elif a==9999 :
    print("Value of a is 9999")
else:
    print("Value of a is Something else")
print("Out of the condtional block")
```

```python
# if Condition = False , elif = True
a = 9999
b = 20
if a==10:# False
    print("value of a is 10")
elif a==9999 : # True
    print("Value of a is 9999")
else:
    print("Value of a is Something else")
print("Out of the condtional block")
```

```python
# if Condition = False , elif = True
a = 5
b = 20
if a==10:# False
    print("value of a is 10")
elif a==9999 : # False
    print("Value of a is 9999")
else:
    print("Value of a is Something else")
print("Out of the condtional block")
```
