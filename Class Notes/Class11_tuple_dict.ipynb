# Properties of a Dictionary
- It is a collection of key and vale pair
- a key and value pair is called as item
- key and value are seperated by (:)
- items are seperated by (,)
- key can not be duplicate , but values can be
- keys should be immutable data types, values can be any data type
- it is not a seq data
- operations --> concatination , membership, identity , repetition
