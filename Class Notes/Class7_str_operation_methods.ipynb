```python
# isalnum ----> it returns True if str is either an alphabet or a number false Otherwise
s1 = "abc" # True
s2 = "123abc" # True
s3 = "**" # False
s4 = "1234" # True
print(s1.isalnum())
print(s2.isalnum())
print(s3.isalnum())
print(s4.isalnum())
```

```python
# isalpha ----> it returns True if str is  an alphabet  false Otherwise
s1 = "abc" # True
s2 = "123abc" # False
s3 = "**" # False
s4 = "1234" # False
print(s1.isalpha())
print(s2.isalpha())
print(s3.isalpha())
print(s4.isalpha())
```

```python
# isascii
s1 = "abc" # True
s2 = "123abc" 
s3 = "**" 
s4 = "1234" 
s5 ="हिन्दी" # False
print(s1.isascii())
print(s2.isascii())
print(s3.isascii())
print(s4.isascii())
print(s5.isascii())
```
