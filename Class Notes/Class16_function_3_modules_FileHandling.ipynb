# filter
- filter is a funct which returns output ONLY if the specified condition is True

```python
syntax
            filter(function , iterable)
```

```python
lst
```

```python
filter(lambda x: x%2==0 , lst)
```

```python
list(filter(lambda x: x%2==0 , lst))
```

```python
emails = ["usman@gmail.com" , "alex@gmail.com", "mamta@yahoo.in", "pavan@hotmail.com"]
```

```python
google_cust = list(filter(lambda x:"@gmail.com" in x, emails ))
```

```python
in
not
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(google_cust , columns=["google_email"])
```
