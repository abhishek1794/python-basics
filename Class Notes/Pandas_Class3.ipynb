```python
df1 = pd.DataFrame({
    "city":["Mumbai" , "Delhi" , "Banglore"],
    "temp":[32,23,30],
    "humidity":[80,90,60]
})
```

# Types of Joins

```python
# Merge ----> to implement joinssyntax
pd.merge(A ,B , on="column" , how="type of join")
```

# inner join

```python
# differnt values
inner_1 = pd.merge(df1 ,df2, on="city" , how="inner")
```

```python
inner_1
```

```python
# common value
inner_2 = pd.merge(df1 ,df3, on="city" , how="inner")
```

```python
inner_2
```

# left Join

```python
# differnt values
left_1 = pd.merge(df1 ,df2, on="city" , how="left")
left_1
```

```python
# common value
left_2 = pd.merge(df1 ,df3, on="city" , how="left")
left_2
```

# Right join

```python
# differnt values
right_1 = pd.merge(df1 ,df2, on="city" , how="right")
right_1
```

```python
# common value
right_2 = pd.merge(df1 ,df3, on="city" , how="right")
right_2
```

# Full Join or Outer Join

```python
# differnt values
full_1 = pd.merge(df1 ,df2, on="city" , how="outer")
full_1
```

```python
# common value
full_2 = pd.merge(df1 ,df3, on="city" , how="outer")
full_2
```
