```python
num1 = eval(input("Enter a first number : "))
operator = input("""
                    + -----> addition
                    - -----> Subtraction
                    * -----> Multiplication
                    / -----> float division
                    // -----> floor division
                     %-----> Modulus
                    ** -----> Power
                  """)
num2 = eval(input("Enter a second number  : "))

if operator=="+":
    print(num1+num2)
elif operator=="-":
    print(num1-num2)
elif operator=="*":
    print(num1*num2)
elif operator=="/":
    print(num1/num2)
elif operator=="//":
    print(num1//num2)
elif operator=="%":
    print(num1%num2)
elif operator=="**":
    print(num1**num2)
else:
    print("Operator not defined")
```
