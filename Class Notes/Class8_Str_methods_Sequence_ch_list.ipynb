```python
username="usman@123"
pasword = "Password"
input_un = input("Enter the user name")
input_password = input("Enter the Password")
if username == input_un.strip() and pasword==input_password.strip():
    print(f"Welcome {username[:5:]}")
else:
    print("Login failed")
```

```python
username="Mohammedusman@123"
pasword = "Password"
input_un = input("Enter the user name")
input_password = input("Enter the Password")
if username == input_un.strip() and pasword==input_password.strip():
    print(f"Welcome {username.split("@")[0]}")
else:
    print("Login failed")
```
