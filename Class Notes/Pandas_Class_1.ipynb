# 2 Creating DF using a dict

```python
d = {"year":[1990, 1994,1998,2002],
    "country":["Italy" , "USA" , "France" , "Japan"],
    "winner":["Germany" , "Brazil" , "France" , "India"],
    }
```

```python
d
```

```python
pd.DataFrame(d)
```

```python
# Creating DF using a Series
my_d = {"year":[1990, 1994,1998,2002,2004],
    "country":["Italy" , "USA" , "France" , "Japan"],
    "winner":["Germany" , "Brazil" , "France" , "India"],
    }
```

```python
my_d
```

```python
pd.DataFrame(my_d)
```

```python
my_d = {"year":pd.Series([1990, 1994,1998,2002,2004]),
    "country":pd.Series(["Italy" , "USA" , "France" , "Japan"]),
    "winner":pd.Series(["Germany" , "Brazil" , "France" , "India"]),
    }
```

```python
pd.DataFrame(my_d)
```
