# Metacharacters
- these are the special characters in Regex which carry some special meaning and are used for a specific purpose

```python
[] - Set
^ - carat
$ - dollar
. - dot
() - group
| - or

```

```python
# [] - Set
- [A] - it will match all A in the string
- [A-Z] - it will match all char from A to Z one by one
- [a-z] - it will match all char from a to z one by one
- [0-9] - it will match all char from 0 to 9 one by one
- [A-Za-z0-9] - it will match all char from  A to Z ,a to z and 0 to 9 one by one
- [^A-Za-z0-9] - it will match all char except  A to Z ,a to z and 0 to 9 one by one
```

# | - or
- abc | xyz ----> either abc or xyz

```python
text = "abc.com xyz.com jkl.com xyz.in mno.co info.com abc.in abc.co"
pat = r"abc|xyz"
x = re.finditer(pat , text)
for i in x:
    print(i)
```

# ( ) - group
- it is used to group the pattern

```python
text = "abc.com xyz.com jkl.com xyz.in mno.co info.com abc.in abc.co"
pat = r"(abc|xyz).com"
x = re.finditer(pat , text)
for i in x:
    print(i)
```

```python
text = "abc.com xyz.com jkl.com xyz.in mno.co info.com abc.in abc.co"
pat = r"(abc|xyz).(com|in)"
x = re.finditer(pat , text)
for i in x:
    print(i)
```

# Quantifiers

{2} --> range - it matche two repetitions in the string
{3,5} ---> it matches repetition in range of 3 to 5
? ---> it matches 1 or no repetition
* ---> it matches single char or group of char or no group
+ ---> it matches single char or group of char or no group

```python
text = "Joy Amith go happy to sleep"
pat = r"[a-z]"
x = re.finditer(pat , text)
for i in x:
    print(i)
```

```python
text = "Joy Amith go happy to sleep"
pat = r"[a-z]{2}"
x = re.finditer(pat , text)
for i in x:
    print(i)
```

```python
text = "Joy Amith go happy to sleep"
pat = r"[a-z]{3}"
x = re.finditer(pat , text)
for i in x:
    print(i)
```

```python
# {3,5}
```

```python
text = "12 5486 22897 DT4565ab 32$ a1 9# 88775241"
pat = r"[0-9]{2}"
x = re.finditer(pat , text)
for i in x:
    print(i)
```

# ? ---> it matches 1 or no repetition
- if there is a match it will return a match object
- if there is a no match it , it will return empty str

```python
text = "12 5486 22897 DT4565ab 32$ a1 9# 88775241"
pat = r"[A-Z]?"
x = re.finditer(pat , text)
for i in x:
    print(i)
```

```python
""
""
""
"D"
```

```python
text = "12 5486 22897 DT4565ab 32$ a1 9# 88775241"
pat = r"[0-9]?"
x = re.finditer(pat , text)
for i in x:
    print(i)
```

# * ---> it matches single char or group of char or no group
- it will find a group of char and return that group
- if no group of char is found it  return that empty str

```python
text = "12 5486 22897 DT4565ab 32$ a1 9# 88775241"
pat = r"[0-9]*"
x = re.finditer(pat , text)
for i in x:
    print(i)
```

```python
12
""
5486
""
22897
""
""
""
4565
```

```python
# + ---> it matches single char or group of char or no group
- it will find a group of char and return that group
- if no group of char is found it  return nothing
```

```python
text = "12 5486 22897 DT4565ab 32$ a1 9# 88775241"
pat = r"[0-9]+"
x = re.finditer(pat , text)
for i in x:
    print(i)
```

```python
12
5486
```
