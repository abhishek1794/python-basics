```python
data type
```

```python
l1 = ["one" , 1 , "two" , 5 , "three" ,"four", 2+3j]
str_value = []
num = []
for i in l1:
    if type(i) == str:
        str_value.append(i)
    elif type(i) == int:
        num.append(i)
    else:
       pass
print(str_value)
print(num)
```

```python
 # l1 = ["one" , 1 , "two" , 5 , "three" ,"four", 2+3j]
def segregator(lst):
    str_value = []
    num = []
    for i in lst:
        if type(i) == str:
            str_value.append(i)
        elif type(i) == int:
            num.append(i)
        else:
           pass
    return str_value,num
```

```python
segregator(["one" , 1 , "two" , 5 , "three" ,"four", 2+3j])
```

```python
def factorial(num):
    fact=1
    for i in range(1, num+1):
       fact=i*fact
    print("Factorial of", num, "is:", fact)
```
