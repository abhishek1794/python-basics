# copy

```python
1- Shallow copy 
2 - Deep copy
```

```python
a = [1,2,3,4,5,6]
b = [1,2,3,4,5,6]

print(id(a))
print(id(b))
```

```python
# aliasing
a = [1,2,3,4,5,6]
b = a

print(id(a))
print(id(b))
```

```python
a[0]
```

```python
a[0] = 99
```

```python
a
```

```python
b
```

```python
# Shallow copy
a = [1,2,3,4,5,6]
b = a.copy()

print(a)
print(b)
```

```python
print(id(a))
print(id(b))
```

```python
# changes in b
b[0] = 9999
```

```python
print(a)
print(b)
```

```python
# failure point of Shallow copy - nested list
a = [1,2,3,4,5,6,[7,8,9]]
b = a.copy() # creating a shallow copy

print("before transformation: ", a)
print("before transformation: ", b)
b[-1][-1] = 1000 # change in b
print("after transformation: ", a)
print("after transformation: ", b)
```

```python
# Shallow copy
a = [1,2,3,4,5,6,[7,8,9]]
b = a.copy() # creating a shallow copy

print(id(a))
print(id(b))
```

```python
print(id(a[-1]))
print(id(b[-1]))
```

```python
from copy import deepcopy
```

```python
# Deep copy
a = [1,2,3,4,5,6,[7,8,9]]
b = deepcopy(a) # creating deep copy

print("before transformation: ", a)
print("before transformation: ", b)
b[-1][-1] = 1000 # change in b
print("after transformation: ", a)
print("after transformation: ", b)
```

```python
a = [1,2,3,4,5,6,[7,8,9]]
b = deepcopy(a) # creating deep copy

print(id(a))
print(id(b))
```

```python
print(id(a[-1]))
print(id(b[-1]))
```
