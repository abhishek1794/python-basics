# 3 - Variable length argument
- *argrs ----> returns a tuple
- **kwargs ----> dict

```python
def add(*args):
    print(args)
    print(type(args))
```

```python
add(1,2,3,4,5,6,7)
```

```python
def add1(*something):
    print(something)
    print(type(something))
add1(4,5,6,7)
```

```python
def add_kw(**kwargs):
    print(kwargs)
    print(type(kwargs))
```

```python
add_kw(a = 10 , b = 20)
```

```python
def add(num1, num2):
    return num1 + num2
def subtract(num1, num2):
    return num1 - num2
```

```python
sign = input("enter a sign: ")
num1 = int(input())
num2 = int(input())
if sign=="+":
    print(add(num1,num2))
elif sign=="-":
    print(subtract(num1,num2))
```
