# Statistical Functions

```python
Mean - average
Median - even - mid values/2 , odd no of values mid value
Mode -  most occuring ? highest frequency
std - deviation of a value from mean
var - square of STD
```

```python
import numpy as np
```

```python
arr = np.array([35,65,17,58,29,45,27,4])
```

```python
arr
```

```python
arr.mean()
```

```python
print(np.mean(arr))
```

```python
(np.median(arr))
```

```python
np.mode(arr)
```

```python
std1 = np.std(arr)
std1
```

```python
np.var(arr)
```

```python
v = std1**2
```

```python
v
```
