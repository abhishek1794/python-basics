# findall()
- it matches all the patterns from the string and retuns a list
- if the pattern if found at it returns a list which contains all the occurances
- if the pattern if not found at it returns empty list


```python
text = "India is my country , India is a great Nation"
pattern = "India"
var = re.findall(pattern , text)
print(var)
```

# finditer()
- it matches all the patterns from the string and retuns a match object
- if the pattern if found at it returns match object which contains all the occurances
- if the pattern if not found at it returns empty iter object
- we need to loop to get match object


```python
text = "India is my country , India is a great Nation"
pattern = "India"
var = re.finditer(pattern , text)
print(var)
```

```python
for i in var:
    print(i)
```

```python
text = "India is my country , India is a great Nation"
pattern = "Iran"
var = re.finditer(pattern , text)
print(var)
```

```python
for i in var:
    print(i)
```
